# Phense — Notebook 01: Data Exploration
**Goal:** Understand the dataset structure, distributions, and class balance before any preprocessing.

Run this notebook after placing raw data files in `data/raw/`.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('YlOrBr')
print('Libraries loaded.')
```

## 1. Load Raw Data
Point to your downloaded file below.

```python
# TODO: Update this path to your actual downloaded file
RAW_FILE = '../data/raw/YOUR_FILE_HERE.csv'

df = pd.read_csv(RAW_FILE)
print(f'Shape: {df.shape}')
df.head(10)
```

## 2. Basic Info

```python
df.info()
print('\nMissing values:')
print(df.isnull().sum())
```

## 3. Numeric Distributions

```python
numeric_cols = ['rainfall_mm', 'elevation_m', 'river_distance_km']
df[numeric_cols].describe()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, col in zip(axes, numeric_cols):
    df[col].hist(bins=30, ax=ax, color='#8B5E3C', edgecolor='white')
    ax.set_title(col)
plt.tight_layout()
plt.savefig('../data/processed/eda_distributions.png', dpi=120)
plt.show()
```

## 4. Class Balance

```python
if 'risk_class' in df.columns:
    counts = df['risk_class'].value_counts()
    print(counts)
    counts.plot(kind='bar', color=['#E0C29A','#8B5E3C','#6F4627','#34302C'], edgecolor='white')
    plt.title('Risk Class Distribution')
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.savefig('../data/processed/eda_class_balance.png', dpi=120)
    plt.show()
else:
    print('risk_class column not found — check column mapping.')
```
